```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
centers = np.array([
    [221, 120],
    [218, 109],
    [209, 112],
    [215, 100],
    [222, 118],
    [212, 120]
])

forwards = np.array([
    [200, 96],
    [202, 102],
    [210, 100],
    [194, 100],
    [211, 99],
    [218, 101]
])

guards = np.array([
    [190, 88],
    [197, 93],
    [180, 75],
    [186, 90],
    [188, 87],
    [193, 90]
])

print(centers.shape[0], forwards.shape[0], guards.shape[0])
```

```python
categories =np.array(["centers", "forwards", "guards"])
print(categories)
```

```python
players =np.stack((centers, forwards, guards))
print(players.shape)
```

```python
print("Maksimipituus:", np.max(players[:, :, 0]))
print("Maksimipaino:", np.max(players[:, :, 1]))
```

```python
print("Keskipituudet:")
print(np.mean(players[:, :, 0], axis=1))
```

```python
for i in range(len(categories)):
    plt.scatter(
        players[i, :, 0],
        players[i, :, 1],
        label=categories[i]
    )

plt.xlabel("Pituus")
plt.ylabel("Paino")
plt.title("Pelaajien pituudet ja painot")
plt.legend()
plt.show()
```

```python
plt.boxplot(
   [centers[:, 0], forwards[:, 0], guards[:, 0]]
)


plt.xlabel("Pelaajakategoria")
plt.ylabel("Pituus")
plt.title("Pelaajien pituudet kategorioittain")
plt.show()
```

```python
x = np.linspace(0, 10, 1000)

y1 = x**2 * np.sin(6 * x)
y2 = np.exp(x) / 250
y3 = 20 * x - 75

plt.plot(x, y1, label="y = x² sin(6x)")
plt.plot(x, y2, label="y = exp(x) / 250")
plt.plot(x, y3, label="y = 20x - 75")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Funktioiden kuvaajat")
plt.legend()
plt.grid(True)
plt.show()
```
