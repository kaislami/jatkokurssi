```python
import numpy as np
```

```python
a = np.array([1, 2, 3, 4])
print("arvosanat: ", a)
```

```python
op = np.array([4, 3, 2, 1])
print("opintopisteet: ", op)
```

```python
print("Kerskiarvo: ", np.average(a))
print("Painotettu kerskiarvo: ", np.average(a, weights=op))
```

```python
print("myös pain.ka.", np.sum(a*op) / np.sum(op))
```

```python
print("myös ka pa", np.matmul(a, op) / np.sum(op))
```
